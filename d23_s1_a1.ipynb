# Day 23 - GAN Loss Functions
Implement discriminator loss and the saturating/non-saturating generator losses using NumPy probabilities.

```python
import numpy as np

d_real = np.array([0.9, 0.8, 0.7])
d_fake = np.array([0.2, 0.3, 0.4])
```

```python
def d_loss(d_real, d_fake):
    loss_real = -np.mean(np.log(d_real))
    loss_fake = -np.mean(np.log(1 - d_fake))
    return float(loss_real + loss_fake)

def g_loss_nonsaturating(d_fake):
    return float(-np.mean(np.log(d_fake)))

def g_loss_saturating(d_fake):
    return float(np.mean(np.log(1 - d_fake)))
```

```python
print(round(d_loss(d_real, d_fake), 4))
print(round(g_loss_nonsaturating(d_fake), 4))
print(round(g_loss_saturating(d_fake), 4))

sure_fake = np.array([1e-4, 1e-4, 1e-4])
print(round(g_loss_saturating(sure_fake), 4))
print(round(g_loss_nonsaturating(sure_fake), 4))
```
