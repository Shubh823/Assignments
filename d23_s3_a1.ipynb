Nearest-neighbour fidelity distance and mode coverage.

```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.neighbors import NearestNeighbors

centers = np.array([[0., 0.], [5., 0.], [0., 5.], [5., 5.]])
real, _ = make_blobs(n_samples=400, centers=centers, cluster_std=0.3, random_state=0)
gen, _ = make_blobs(n_samples=400, centers=centers, cluster_std=0.3, random_state=1)
```

```python
def nn_distance(real, gen):
    nn = NearestNeighbors(n_neighbors=1).fit(real)
    dists, _ = nn.kneighbors(gen)
    return float(dists.ravel().mean())

def mode_coverage(real_centers, gen, radius):
    nn = NearestNeighbors(n_neighbors=1).fit(gen)
    dists, _ = nn.kneighbors(real_centers)
    return float(np.mean(dists.ravel() <= radius))
```

```python
# Verification
real_test = np.random.RandomState(0).randn(40, 2)
print(round(nn_distance(real_test, real_test), 6))
grid = np.array([[i*3., j*3.] for i in range(5) for j in range(5)])
print(round(nn_distance(grid, grid + np.array([0., 0.5])), 3))

rng = np.random.RandomState(0)
gen_good = np.vstack([c + 0.1*rng.randn(30, 2) for c in centers])
gen_bad = centers[0] + 0.1*rng.randn(120, 2)
print(mode_coverage(centers, gen_good, radius=1.0))
print(mode_coverage(centers, gen_bad, radius=1.0))
```
