# Day 23 Session 2 Assignment 2 - VAE Diagnostics and beta-VAE
Per-dimension KL diagnostics and beta-weighted VAE objective using NumPy.

```python
import numpy as np

rng = np.random.default_rng(1)
N = 500
mu = np.zeros((N, 6)); logvar = np.zeros((N, 6))
mu[:, 0] = rng.normal(0, 2.0, size=N)
mu[:, 1] = rng.normal(0, 1.5, size=N)
logvar[:, 2] = 1.5
```

```python
def per_dim_kl(mu, logvar):
    kl_each = 0.5 * (np.exp(logvar) + mu**2 - 1 - logvar)
    return np.mean(kl_each, axis=0)

def beta_elbo(recon, kl, beta):
    return recon + beta * kl
```

```python
pk = per_dim_kl(mu, logvar)
print(np.round(pk, 3))
print(np.round(pk.sum(), 4))

print(beta_elbo(2.0, 3.0, 1.0))
print(beta_elbo(2.0, 3.0, 4.0))
print(beta_elbo(2.0, 3.0, 0.0))
```
