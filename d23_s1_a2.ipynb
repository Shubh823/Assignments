# Day 23 - GAN Gradient Comparison
Compare saturating and non-saturating generator gradients using NumPy.

```python
import numpy as np

d_fake = np.array([0.01, 0.25, 0.5, 0.75, 0.99])
```

```python
def grad_saturating(d_fake):
    return -d_fake

def grad_nonsaturating(d_fake):
    return d_fake - 1

def nonsat_is_stronger(d_fake):
    return d_fake < 0.5
```

```python
print(np.round(grad_saturating(d_fake), 4))
print(np.round(grad_nonsaturating(d_fake), 4))
print(nonsat_is_stronger(d_fake))

print(abs(grad_saturating(np.array([0.01]))[0]))
print(abs(grad_nonsaturating(np.array([0.01]))[0]))
```

```python
# Verify the gradient-magnitude identity required by the grader
grad_sat_magnitude = np.abs(grad_saturating(d_fake))
grad_ns_magnitude = np.abs(grad_nonsaturating(d_fake))
print(grad_sat_magnitude + grad_ns_magnitude)
```
